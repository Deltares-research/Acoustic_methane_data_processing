```python


import datetime

import glob
import os
import cv2
import matplotlib.pyplot as plt
from datetime import datetime
output_path = r"P:\11212489-008-acoustic-ch4-test\experiment_data\261021_wc_experiment_01\261021_wc_exp_01_3\261021_wc_exp_01_3"
img_files = sorted(glob.glob(os.path.join(output_path, "*.bmp")))

# retrieve the time stamps from the image file names

def get_timestamp_from_filename(filename):

    date_ = filename.split('\\')[-1].split('_')[0]
    yyyy = '20'+date_[:2]
    mm = date_[2:4]
    dd = date_[4:6]

    time_ = filename.split('\\')[-1].split('-')[-1]
    hh = time_[:2]
    mi = time_[3:5]
    ss = time_[6:8]

    return datetime(int(yyyy), int(mm), int(dd), int(hh), int(mi), int(ss))
```

```python
output_path = r"P:\11212489-008-acoustic-ch4-test\data_processing\images_files_261021_wc_experiment_03"

# make sure the output path exists
os.makedirs(output_path, exist_ok=True)


# read each image file and store them in a list
for f in img_files:

    image = cv2.imread(f, cv2.IMREAD_GRAYSCALE)
    # rotate the image 90 degrees clockwise
    image = cv2.rotate(image, cv2.ROTATE_90_CLOCKWISE)

    # read time information from the filename
    timestamp = os.path.splitext(os.path.basename(f))[0]

    # store the image and timestamp in a dictionary
    image_data = {"timestamp": timestamp, "image": image}

    # save each image as png with smaller size
    save_path = os.path.join(output_path, f"{timestamp}.png")
    cv2.imwrite(save_path, image)
```
